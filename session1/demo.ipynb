What does this mean? First, let me reframe these results: 

```python
ct.groupby(['visitor_type','channel'])['converted'].mean().unstack()
```

If we look at just "new" visitors - Paid Social performs better (6% vs 4% conversion). And if we look at just "returning" visitors, Paid Social ALSO performs better (20% vs 14%). But when we looked at the crude data, it looks like retargeting performed better (12% vs 7.4%). 

Why did this happen? To understand this better, let's revisit the code the AI suggested (above)

```python
ct.groupby(['channel', 'visitor_type'])['converted'].agg(['sum','count','mean'])
```

Retargeting looks efficient because 80% of its traffic is people who already buy from us. It isn't converting better. It's fishing in a barrel we already own.

Notice that if we looked at spend data, Paid Social also comes out as the winner: 

```python
spend = pd.read_csv('channel_spend.csv')
spend
```

```python
df = ct.groupby(['channel'])['converted'].agg(['sum','count','mean']).merge(spend, how='inner', on='channel')
df
```
